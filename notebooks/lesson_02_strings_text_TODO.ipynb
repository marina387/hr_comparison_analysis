## Краткая теория по ноутбуку

На этом занятии мы учимся работать с **текстовыми данными**.  
В аналитике и в системах с ИИ текст встречается постоянно:

- названия товаров
- категории
- коды
- ФИО
- названия компаний
- e-mail и теги
- данные из форм, CRM и Excel

### Что важно понять
Текст в Python хранится в типе данных **`str`**.  
Со строками можно:

- получать длину текста
- извлекать отдельные символы
- брать фрагменты строки
- очищать лишние пробелы
- менять регистр
- заменять символы
- объединять несколько строк в одну

### Базовые термины
- **Строка** — текстовое значение
- **Индекс** — номер символа в строке
- **Срез** — часть строки
- **Метод строки** — встроенная операция, которую можно применить к тексту

### Что изучаем сегодня
1. Создание строк и измерение длины текста

4. Склейку строк и форматирование текста
5. Мини-задачу по очистке текстовых данных

### Связь с Excel
Сегодня мы также увидим аналоги в Excel:

- `LEN()` ↔ `len()`
- `LEFT()/RIGHT()/MID()` ↔ индексация и срезы
- `TRIM()` ↔ `.strip()`
- `LOWER()/UPPER()` ↔ `.lower() / .upper()`
- `SUBSTITUTE()` ↔ `.replace()`
- `CONCAT()` ↔ склейка строк

## Практическая ячейка 1. Строки и длина текста

### Что делаем
Создаём текстовые переменные и измеряем длину строки.

### Функции, методы и синтаксис
- `"текст"` — строковое значение
- `len(строка)` — длина строки
- `print(...)` — вывод результата

### Задание
1. Выведите исходные значения.
2. Посчитайте длину строки `client_name`.
3. Посчитайте длину строки `product_code`.

```python
import pandas as pd

# Загрузка наших HR-датасетов
# Из папки notebooks/
df_hf = pd.read_csv("../data/huggingface/resume_scores_clean.csv")
df_kaggle = pd.read_csv("../data/kaggle_resume/resume_data.csv")
df_hh = pd.read_csv("../data/hhru/it.csv", sep=';')

print("✅ Данные загружены")
print(f"Hugging Face: {df_hf.shape}")
print(f"Resume Dataset: {df_kaggle.shape}")
print(f"hh.ru: {df_hh.shape}")

# Берём несколько примеров из датасетов
sample_name = df_hf['name'].iloc[0] if 'name' in df_hf.columns else "Ирина Смирнова"
sample_skills = df_hf['skills'].iloc[0] if 'skills' in df_hf.columns else "Python, SQL, AWS"
sample_job = df_hh['name'].iloc[0] if 'name' in df_hh.columns else "Аналитик данных"

print(f"\nПример имени из датасета: {sample_name}")
print(f"Пример навыков из датасета: {sample_skills}")
print(f"Пример вакансии из hh.ru: {sample_job}")
```

```python
# Берём реальные данные из датасетов
client_name = sample_name
skills_text = sample_skills
job_title = sample_job

# Вычисляем длину
client_name_length = len(client_name)
skills_length = len(skills_text)

print("Исходное имя кандидата:", repr(client_name))
print(f"Длина имени: {client_name_length} символов")
print(f"Навыки: {skills_text}")
print(f"Длина строки навыков: {skills_length} символов")
print(f"Название вакансии: {job_title}")
print(f"Длина названия вакансии: {len(job_title)} символов")
```

## Практическая ячейка 2. Индексация и срезы

### Что делаем
Получаем отдельные символы и части строки.

### Функции, методы и синтаксис
- `строка[0]` — первый символ
- `строка[-1]` — последний символ
- `строка[0:2]` — срез
- `строка[-4:]` — последние 4 символа

### Задание
1. Найдите первый символ.
2. Найдите последний символ.
3. Получите префикс `AI`.
4. Получите последние 4 символа года.

```python
# Берём строку с навыками
skills = skills_text

# Индексация
first_skill_char = skills[0] if len(skills) > 0 else ""
last_skill_char = skills[-1] if len(skills) > 0 else ""

# Срезы
first_skill = skills.split(',')[0].strip() if ',' in skills else skills[:5]
last_five_chars = skills[-5:] if len(skills) >= 5 else skills

print(f"Навыки: {skills}")
print(f"Первый символ: {first_skill_char}")
print(f"Последний символ: {last_skill_char}")
print(f"Первый навык: {first_skill}")
print(f"Последние 5 символов строки навыков: {last_five_chars}")

# Дополнительно: извлечение языка программирования из строки
if 'Python' in skills:
    print("✅ Кандидат знает Python")
else:
    print("❌ Python не указан в навыках")
```

## Практическая ячейка 3. Очистка текста и методы строк

### Что делаем
Очищаем строку от лишних пробелов, меняем регистр и заменяем символы.

### Функции, методы и синтаксис
- `.strip()`
- `.lower()`
- `.upper()`
- `.replace("старое", "новое")`

### Задание
1. Уберите пробелы с краёв.
2. Переведите строку в нижний регистр.
3. Переведите строку в верхний регистр.
4. Замените `-` на пробел.

```python
# Берём "грязные" данные из датасетов (специально создаём примеры)
raw_job_title = "  Аналитик-данных-ПРОМЫШЛЕННОСТЬ  "
raw_skill = "  python, SQL,  MACHINE LEARNING  "
raw_company = "  техно-СЕРВИС  "

# Очистка
clean_job_title = raw_job_title.strip()
lower_job_title = clean_job_title.lower()
upper_job_title = clean_job_title.upper()
replaced_job_title = clean_job_title.replace("-", " ")

clean_skill = raw_skill.strip()
normalized_skill = clean_skill.lower().replace(",", ", ")

print("Исходная вакансия:", repr(raw_job_title))
print("После strip():", repr(clean_job_title))
print("После lower():", lower_job_title)
print("После upper():", upper_job_title)
print("После replace():", replaced_job_title)

print("\nИсходные навыки:", repr(raw_skill))
print("После очистки:", clean_skill)
print("Нормализованные навыки:", normalized_skill)
```

## Практическая ячейка 5. Мини-задача: очистка карточки товара

### Что делаем
Берём «грязные» текстовые данные и превращаем их в аккуратные значения для анализа.

### Функции, методы и синтаксис
- `.strip()`
- `.lower()`

### Задание
1. Очистите название товара.

3. Сделайте категорию в нижнем регистре.

```python
# Берём реальные "грязные" данные (создаём примеры на основе датасетов)
raw_vacancy = "  Python-разработчик-Сеньор  "
raw_skills = "  python,  django,  POSTGRESQL,  docker  "
raw_format = "  Удаленная-работа/полный-день  "

# Очистка
vacancy_clean = raw_vacancy.strip().lower().replace("-", " ")
skills_clean = raw_skills.strip().lower().replace("  ", " ")
format_clean = raw_format.strip().replace("-", " / ")

# Формирование карточки вакансии
job_card = f"""
ВАКАНСИЯ
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
Должность: {vacancy_clean.title()}
Требуемые навыки: {skills_clean}
Формат работы: {format_clean}
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
"""

print("Очищенное название вакансии:", vacancy_clean)
print("Очищенные навыки:", skills_clean)
print("Формат работы:", format_clean)
print("\nКарточка вакансии:")
print(job_card)
```
